```python
import json
from pathlib import Path
import sys

# Ensure repository root is on sys.path so absolute imports work in Jupyter
repo_root = Path.cwd()
if (repo_root / "src").exists() and str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# Locate repository root by walking up until a folder containing "src" is found
def find_repo_root(start: Path = repo_root, max_levels: int = 6) -> Path:
    p = start
    for _ in range(max_levels):
        if (p / "src").exists():
            return p
        if p.parent == p:
            break
        p = p.parent
    return start

repo_root = find_repo_root(repo_root)
if (repo_root / "src").exists() and str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.core.base import *
from src.core.filemanager import *
```
